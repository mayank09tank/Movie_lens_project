```python
def student_management_system():
    students = []

    def add_student(name, age, email, phone_number):
        student = {
            'name': name,
            'age': age,
            'email': email,
            'phone_number': phone_number
        }
        students.append(student)  
        print(f"Student {name} added successfully.")

    def view_students():
        if not students:
            print("No students found.")
            return
        for student in students:
            print(f"Name: {student['name']}, Age: {student['age']}, Email: {student['email']}, Phone: {student['phone_number']}")  
    def remove_student(name):
        nonlocal students
        students = [s for s in students if s['name'] != name]  
        print(f"Student {name} removed successfully.")

    while True:  
        print("\nStudent Management System")
        print("1. Add Student")
        print("2. View Students")
        print("3. Remove Student")
        print("4. Exit")  

        choice = input("Enter your choice: ")

        if choice == '1':
            name = input("Enter student's name: ")
            age = int(input("Enter student's age: "))
            email = input("Enter student's email: ")
            phone_number = input("Enter student's phone number: ")
            add_student(name, age, email, phone_number)

        elif choice == '2':
            view_students()

        elif choice == '3':
            name = input("Enter student's name to remove: ")
            remove_student(name)

        elif choice == '4':
            print("Exiting the system.")
            break

        else:
            print("Invalid choice. Please try again.")

student_management_system()
```
