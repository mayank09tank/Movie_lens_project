```python
# atm withdrawl
pin = input("Enter your pin: ")
balance = int(input("Enter your balance: "))
amount = int(input("Enter the amount to withdraw: "))
# first chek pin is correct or not
#if correct,check weather amount us positve.
# if positive , check weather sufficent balance is available
# if available , process the withdrawal
if pin == '1234':
    print('pin is correct')
    if amount > 0:
        print('amount is valid')
        if amount <= balance:
            print('withdrawal processed')
            print('remaining balance:', balance-amount)
        else:
            print('insufficient balance')
    else:
        print('invalid amount')
```
